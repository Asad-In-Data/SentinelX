```python
import pandas as pd
import os 


df = pd.read_csv(r"C:\Users\Asad Ali\Desktop\SentinelX\network_traffic.csv")

print("Total Packets:", len(df))
print("\nProtocol Distribution:")
print(df["Protocol"].value_counts())

print("\nTop Source IPs:")
print(df["Source IP"].value_counts().head(5))

print("\nTop Destination IPs:")
print(df["Destination IP"].value_counts().head(5))
```

```python
import matplotlib.pyplot as plt

df["Protocol"].value_counts().plot(kind='bar')
plt.title("Protocol Distribution")
plt.xlabel("Protocol")
plt.ylabel("Count")
plt.show()
```
